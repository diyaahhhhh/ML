```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
df=pd.read_excel("Online Retail.xlsx")        #loading dataset
print(df.head())     #displaying first five records
print(df.info())         #dataset information
```

```python
df=df.dropna(subset=["CustomerID"])          #removing customers with missing customer ids
df=df[~df["InvoiceNo"].astype(str).str.startswith("C")]
df=df[df["Quantity"]>0]
df=df[df["UnitPrice"]>0]
df["TotalAmount"]=df["Quantity"]*df["UnitPrice"]
```

```python
print(df.shape)    #provides no. of rows an no. of columns
print(df.head())
```

```python
#grouping customers based on customerID- called customer-level aggregation

customer_data=df.groupby("CustomerID").agg(
    

TotalSpend=("TotalAmount","sum"),
TotalQuantity=("Quantity","sum"),
NumberOfTransactions=("InvoiceNo","nunique"),
NumberOfProducts=("StockCode","nunique"),
AverageOrderValue=("TotalAmount","mean"),
Country=("Country","first")
).reset_index()

print(customer_data.head())
```

```python
customer_data["Segment"]=pd.qcut(                #target variable is Segment
    customer_data["TotalSpend"],
    q=3,
    labels=["Low Value","Medium Value","High Value"]
)
print(customer_data["Segment"].value_counts() )    #value count- how many customers belong to each
```

```python
features=[
    "TotalQuantity",
    "NumberOfTransactions",
    "NumberOfProducts",
    "AverageOrderValue"
]
x=customer_data[features]
y=customer_data["Segment"]
```

```python
x_train, x_test, y_train, y_test=train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
   )
```

```python
id3_model=DecisionTreeClassifier(
    criterion="entropy",
    random_state=42,
    max_depth=4
)
id3_model.fit(x_train, y_train)
y_pred=id3_model.predict(x_test)
id3_model.fit(x_train, y_train)
y_pred=id3_model.predict(x_test)
print("Predictions:")
print(y_pred[:20])
```

```python
accuracy=accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test,y_pred))
```

```python
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
importance=pd.DataFrame({
    "Feature":features,
    "Importance":id3_model.feature_importances_})
importance_sorted=importance.sort_values(
    "Importance",
    ascending=True
)
plt.figure(figsize=(40,36))
```

```python
plot_tree(
    id3_model,
    feature_names=features,
class_names=id3_model.classes_,
filled=True,
rounded=True,
proportion=False,
)
plt.title("Customer Segmentation Using ID3 Decision Tree")
plt.show()
```

```python
plt.figure(figsize=(10,6))
importance_sorted=importance.sort_values(
    "Importance",
    ascending=True
)
plt.barh(
    importance_sorted["Feature"],
    importance_sorted["Importance"]
)
plt.xlabel("Information-based importance")
plt.ylabel("Customer Behaviour Features")
plt.title("Feature Importance in Customer Segmentation")
plt.show()
```
